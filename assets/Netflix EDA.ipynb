```python
import pandas as pd
df = pd.read_csv('netflix_india_yt_data.csv')

# Basic cleaning ONLY (no features - DAX handles)
df = df.drop_duplicates(subset=['video_id'])  # Assumes column name
df.fillna({
    'view_count': 0,
    'like_count': 0,
    'comment_count': 0,
    'tags': '',  # Empty string for DAX
    'duration': 'PT0S',  # Default duration
    'published_date': '2020-01-01'  # Default date
}, inplace=True)

df.to_csv('netflix_ready.csv', index=False)
print("Done")
```
